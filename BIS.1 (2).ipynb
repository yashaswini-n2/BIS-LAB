# week 1 [ GENERIC ALGORITHM]
Employee shift scheduling

```python
import random

employees = 6
shifts = 3
POP, GEN, MUT = 20, 50, 0.1

def fitness(schedule):
    counts = [schedule.count(i) for i in range(shifts)]
    return sum(abs(x - 2) for x in counts)  

population = [
    [random.randrange(shifts) for _ in range(employees)]
    for _ in range(POP)
]

for _ in range(GEN):
    population.sort(key=fitness)
    parents = population[:10]
    new = parents.copy()

    while len(new) < POP:
        p1, p2 = random.sample(parents, 2)
        point = random.randint(1, employees - 1)
        child = p1[:point] + p2[point:]

        if random.random() < MUT:
            child[random.randrange(employees)] = random.randrange(shifts)

        new.append(child)

    population = new

best = min(population, key=fitness)

print("Best Schedule:", best)
print("Conflicts:", fitness(best))
```

# week 2 [PARICLE SWARM OPTIMIZATION ]
[FINDING NEAREST SCHOOL]|

```python
import random
import math
student = (50, 50)
def fitness(x, y):
    return math.sqrt((x - student[0])**2 + (y - student[1])**2)

n = 10
iterations = 30
w = 0.5
c1 = 1.5
c2 = 1.5
particles = []

for i in range(n):
    x = random.uniform(0, 100)
    y = random.uniform(0, 100)
    particles.append({
        "x": x,
        "y": y,
        "vx": random.uniform(-1, 1),
        "vy": random.uniform(-1, 1),
        "pbest": (x, y),
        "best": fitness(x, y) })

gbest = min(particles, key=lambda p: p["best"])
gx, gy = gbest["pbest"]
for _ in range(iterations):
    for p in particles:
        f = fitness(p["x"], p["y"])

        if f < p["best"]:
            p["pbest"] = (p["x"], p["y"])
            p["best"] = f
    gbest = min(particles, key=lambda p: p["best"])
    gx, gy = gbest["pbest"]
    for p in particles:
        r1 = random.random()
        r2 = random.random()
        p["vx"] = (
            w * p["vx"]
            + c1 * r1 * (p["pbest"][0] - p["x"])
            + c2 * r2 * (gx - p["x"]))
        p["vy"] = (
            w * p["vy"]
            + c1 * r1 * (p["pbest"][1] - p["y"])
            + c2 * r2 * (gy - p["y"]) )
        p["x"] += p["vx"]
        p["y"] += p["vy"]

print("Nearest School Location:", (round(gx, 2), round(gy, 2)))
print("Distance:", round(fitness(gx, gy), 2))
```

# week 3 [ OPTIMIZATION VIA GENE EXPRESSION ALGORITHM]|
[DELIVERY ROUTE OPTIMIZATION]

```python
import random

dist = [
    [0,10,15,20,25],
    [10,0,35,25,30],
    [15,35,0,30,20],
    [20,25,30,0,15],
    [25,30,20,15,0]
]

N, POP, GEN, MUT = 5, 20, 100, 0.1

def distance(r):
    return sum(dist[r[i]][r[(i+1) % N]] for i in range(N))

population = []
for _ in range(POP):
    r = list(range(N))
    random.shuffle(r)
    population.append(r)

for _ in range(GEN):
    population.sort(key=distance)
    parents = population[:10]
    new = parents.copy()

    while len(new) < POP:
        p1, p2 = random.sample(parents, 2)
        point = random.randint(1, N-2)
        child = p1[:point] + [x for x in p2 if x not in p1[:point]]

        if random.random() < MUT:
            a, b = random.sample(range(N), 2)
            child[a], child[b] = child[b], child[a]

        new.append(child)

    population = new

best = min(population, key=distance)
print("Best Route:", best)
print("Minimum Distance:", distance(best))
```

# week 4 Ant colony optimization 

```python
import random
def ant_colony_tsp(distance, num_ants=10, iterations=100,
                   alpha=1, beta=2, rho=0.5, Q=100):
    n = len(distance)
    pheromone = [[1.0 for _ in range(n)] for _ in range(n)]
    best_tour = None
    best_distance = float("inf")
    for iteration in range(iterations):
        all_tours = []
        for ant in range(num_ants):
            start = random.randrange(n)
            tour = [start]
            unvisited = list(range(n))
            unvisited.remove(start)
            while unvisited:
                current = tour[-1]
                weights = []
                for city in unvisited:
                    tau = pheromone[current][city] ** alpha
                    eta = (1 / distance[current][city]) ** beta
                    weights.append(tau * eta)
                total = sum(weights)
                probabilities = [
                    weight / total for weight in weights ]
                next_city = random.choices(
                    unvisited,
                    weights=probabilities,
                    k=1
                )[0]
                tour.append(next_city)
                unvisited.remove(next_city)
            tour.append(start)
            length = sum(
                distance[tour[i]][tour[i + 1]]
                for i in range(n)
            )
            all_tours.append((tour, length))
            if length < best_distance:
                best_distance = length
                best_tour = tour[:]
        for i in range(n):
            for j in range(n):
                pheromone[i][j] *= (1 - rho)
        for tour, length in all_tours:
            deposit = Q / length
            for i in range(n):
                a = tour[i]
                b = tour[i + 1]
                pheromone[a][b] += deposit
                pheromone[b][a] += deposit
    return best_tour, best_distance
distance = [
    [0, 10, 15, 20, 18],
    [10, 0, 35, 25, 12],
    [15, 35, 0, 30, 10],
    [20, 25, 30, 0, 14],
    [18, 12, 10, 14, 0]]
tour, length = ant_colony_tsp(distance)
print("Best Tour:", " -> ".join(map(str, tour)))
print("Minimum Distance:", length)
```

# week 5 [ GRAY WOLF OPTIMIZATION ]
WIRELESS COMMUNICATION

```python
import numpy as np
bandwidth = 100
signal_strength = 50
interference = 20
n_wolves = 20
max_iter = 100
dim = 3
wolves = np.random.uniform(0, 10, (n_wolves, dim))

alpha = np.zeros(dim)
beta = np.zeros(dim)
delta = np.zeros(dim)

alpha_score = float("inf")
beta_score = float("inf")
delta_score = float("inf")

def fitness(x):
    transmission_power = x[0]
    channel = x[1]
    network_parameter = x[2]
    signal = signal_strength + transmission_power
    noise = interference + network_parameter
    score = noise / (signal + 1e-6)
    return score
for iteration in range(max_iter):
    for i in range(n_wolves):
        score = fitness(wolves[i])
        if score < alpha_score:
            delta_score, delta = beta_score, beta.copy()
            beta_score, beta = alpha_score, alpha.copy()
            alpha_score, alpha = score, wolves[i].copy()
        elif score < beta_score:
            delta_score, delta = beta_score, beta.copy()
            beta_score, beta = score, wolves[i].copy()
        elif score < delta_score:
            delta_score, delta = score, wolves[i].copy()
    a = 2 - iteration * (2 / max_iter)
    for i in range(n_wolves):
        for j in range(dim):
            X = []
            for leader in [alpha, beta, delta]:
                r1, r2 = np.random.random(2)
                A = 2 * a * r1 - a
                C = 2 * r2
                D = abs(C * leader[j] - wolves[i, j])
                X.append(leader[j] - A * D)
            wolves[i, j] = sum(X) / 3
        wolves[i] = np.clip(wolves[i], 0, 10)
print("Optimal Network Parameters:", alpha)
print("Minimum Interference Fitness:", alpha_score)
print("Transmission Power:", alpha[0])
print("Channel Parameter:", alpha[1])
print("Network Parameter:", alpha[2])
```

# week5 [ CUCKOO SEARCH ALGORITHM ]

```python
import random
jobs = [5, 3, 2, 4]
n = 10
Pa = 0.25
MaxIter = 100
def fitness(schedule):
    machine_time = 0
    max_time = 0
    for job in schedule:
        machine_time += jobs[job]
    return machine_time
def generate_schedule():
    schedule = list(range(len(jobs)))
    random.shuffle(schedule)
    return schedule
nests = [generate_schedule() for _ in range(n)]
best = min(nests, key=fitness)

for t in range(MaxIter):

    for i in range(n):
        new_schedule = generate_schedule()
        j = random.randint(0, n - 1)
        if fitness(new_schedule) < fitness(nests[j]):
            nests[j] = new_schedule
    for i in range(n):
        if random.random() < Pa:
            nests[i] = generate_schedule()
    current_best = min(nests, key=fitness)
    if fitness(current_best) < fitness(best):
        best = current_best.copy()
print("Best Job Schedule:")
for job in best:
    print("Job", job + 1, "->", jobs[job], "hours")
print("Minimum Completion Time:", fitness(best), "hours")
```
