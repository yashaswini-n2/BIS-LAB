# week 1

```python
import numpy as np
def objective_function(x):
    return np.sum(x**2) 

class PSO:
    def __init__(self, n_particles, dim, bounds, max_iter,
                 w=0.5, c1=1.5, c2=1.5):
        self.n_particles = n_particles
        self.dim = dim
        self.bounds = bounds
        self.max_iter = max_iter
        self.w = w     
        self.c1 = c1   
        self.c2 = c2   
        self.positions = np.random.uniform(bounds[0], bounds[1],
                                           (n_particles, dim))
        self.velocities = np.random.uniform(-1, 1,
                                            (n_particles, dim))
        self.pbest_pos = self.positions.copy()
        self.pbest_val = np.array([objective_function(p) for p in self.positions])
        best_idx = np.argmin(self.pbest_val)
        self.gbest_pos = self.pbest_pos[best_idx].copy()
        self.gbest_val = self.pbest_val[best_idx]

    def optimize(self):
        for _ in range(self.max_iter):

            for i in range(self.n_particles):
                fitness = objective_function(self.positions[i])
                if fitness < self.pbest_val[i]:
                    self.pbest_val[i] = fitness
                    self.pbest_pos[i] = self.positions[i].copy()
                    
            best_idx = np.argmin(self.pbest_val)
            self.gbest_pos = self.pbest_pos[best_idx].copy()
            self.gbest_val = self.pbest_val[best_idx]
            for i in range(self.n_particles):
                r1 = np.random.rand(self.dim)
                r2 = np.random.rand(self.dim)

                self.velocities[i] = (
                    self.w * self.velocities[i] +
                    self.c1 * r1 * (self.pbest_pos[i] - self.positions[i]) +
                    self.c2 * r2 * (self.gbest_pos - self.positions[i])  )
                self.positions[i] += self.velocities[i]
                self.positions[i] = np.clip(self.positions[i],
                                            self.bounds[0],
                                            self.bounds[1])
        return self.gbest_pos, self.gbest_val
if __name__ == "__main__":
    pso = PSO(
        n_particles=30,
        dim=2,
        bounds=(-10, 10),
        max_iter=100 )
    best_pos, best_val = pso.optimize()
    print("Best Position:", best_pos)
    print("Best Value:", best_val)
```

# week 2

```python
import random
import math

student = (50, 50)
def fitness(x, y):
    return math.sqrt((x - student[0])**2 + (y - student[1])**2)

n = 10
iterations = 30
w = 0.5
c1 = 1.5
c2 = 1.5
particles = []

for i in range(n):
    x = random.uniform(0, 100)
    y = random.uniform(0, 100)
    particles.append({
        "x": x,
        "y": y,
        "vx": random.uniform(-1, 1),
        "vy": random.uniform(-1, 1),
        "pbest": (x, y),
        "best": fitness(x, y) })

gbest = min(particles, key=lambda p: p["best"])
gx, gy = gbest["pbest"]
for _ in range(iterations):

    for p in particles:
        f = fitness(p["x"], p["y"])

        if f < p["best"]:
            p["pbest"] = (p["x"], p["y"])
            p["best"] = f
    gbest = min(particles, key=lambda p: p["best"])
    gx, gy = gbest["pbest"]
    for p in particles:
        r1 = random.random()
        r2 = random.random()
        p["vx"] = (
            w * p["vx"]
            + c1 * r1 * (p["pbest"][0] - p["x"])
            + c2 * r2 * (gx - p["x"]))
        p["vy"] = (
            w * p["vy"]
            + c1 * r1 * (p["pbest"][1] - p["y"])
            + c2 * r2 * (gy - p["y"]) )
        p["x"] += p["vx"]
        p["y"] += p["vy"]

print("Nearest School Location:", (round(gx, 2), round(gy, 2)))
print("Distance:", round(fitness(gx, gy), 2))
```

# week 3

```python
import random

dist = [
    [0,10,15,20,25],
    [10,0,35,25,30],
    [15,35,0,30,20],
    [20,25,30,0,15],
    [25,30,20,15,0]
]

N, POP, GEN, MUT = 5, 20, 100, 0.1

def distance(r):
    return sum(dist[r[i]][r[(i+1) % N]] for i in range(N))

population = []
for _ in range(POP):
    r = list(range(N))
    random.shuffle(r)
    population.append(r)

for _ in range(GEN):
    population.sort(key=distance)
    parents = population[:10]
    new = parents.copy()

    while len(new) < POP:
        p1, p2 = random.sample(parents, 2)
        point = random.randint(1, N-2)
        child = p1[:point] + [x for x in p2 if x not in p1[:point]]

        if random.random() < MUT:
            a, b = random.sample(range(N), 2)
            child[a], child[b] = child[b], child[a]

        new.append(child)

    population = new

best = min(population, key=distance)
print("Best Route:", best)
print("Minimum Distance:", distance(best))
```
